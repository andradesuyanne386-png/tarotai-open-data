```python
import json
from pathlib import Path
paths = list(Path("/kaggle/input").rglob("cards.json"))
assert len(paths) == 1, "Attach the TarotAI dataset only"
cards = json.loads(paths[0].read_text())
assert len(cards) == 78
from collections import Counter
print(Counter(c["arcana"] for c in cards))
query = "beginnings"
[c for c in cards if query in c["upright"]]
```
